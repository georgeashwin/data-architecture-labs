```python
df= spark.read.csv('dbfs:/FileStore/datasets/authors_publications.csv', inferSchema=True, header=True)
```

```python
display(df)
```

```python
from pyspark.sql.types import IntegerType
df = df.withColumn("citations", df["citations"].cast(IntegerType()))
df_yearly_citations = df.groupBy("year").sum("citations").withColumnRenamed("sum(citations)", "total_citations")

display(df_yearly_citations)
```
